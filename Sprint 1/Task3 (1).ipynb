```
using System;

public class Event                                              // Базовый класс
{
    private int eventId;                                       
    private string eventName;                                   // Инкапсуляция
    private DateTime date;                                      

    public int EventId                                          
    {
        get { return eventId; }                                
        set { eventId = value; }                              
    }

    public string EventName                                     
    {
        get { return eventName; }                               
        set { eventName = value; }                             
    }

    public DateTime Date                                        
    {
        get { return date; }                                    
        set { date = value; }                                  
    }

    public Event()                                              // Конструктор по умолчанию
    {
        EventId = 0;
        EventName = "Unknown";
        Date = DateTime.Now;
    }

    public Event(int id, string name, DateTime date)            // Конструктор с параметрами
    {
        EventId = id;
        EventName = name;
        Date = date;
    }

    public virtual void RegisterParticipant()                   // Полиморфизм: virtual метод
    {
        Console.WriteLine("Регистрация участника на мероприятие");
    }

    public virtual void NotifyParticipants()                    // Полиморфизм: virtual метод
    {
        Console.WriteLine("Уведомление участников о мероприятии");
    }

    public virtual void GetEventDetails()                       // Полиморфизм: virtual метод
    {
        Console.WriteLine($"Мероприятие: {eventName}, ID: {eventId}, Дата: {date}");
    }

    public void InteractWith(Event other)                       // Взаимодействие объектов
    {
        Console.WriteLine($"[{EventName}] взаимодействует с [{other.EventName}]");
        Console.WriteLine("   Мероприятия обмениваются участниками.");
    }
}

public class Conference : Event                                 // Наследование от Event
{
    private int speakersCount;                                  

    public int SpeakersCount                                    
    {
        get { return speakersCount; }                           
        set { speakersCount = value; }                         
    }

    public Conference() : base()                                // Конструктор (вызывает base)
    {
        SpeakersCount = 0;
    }

    public Conference(int id, string name, DateTime date, int speakers) : base(id, name, date) // Конструктор с параметрами
    {
        SpeakersCount = speakers;
    }

    public override void NotifyParticipants()                   // Полиморфизм: override
    {
        Console.WriteLine($"Уведомление участников конференции с программой. Спикеров: {speakersCount}");
    }
}

public class Festival : Event                                   // Наследование от Event
{
    private int bandsCount;                                     

    public int BandsCount                                       // Своё свойство
    {
        get { return bandsCount; }                              
        set { bandsCount = value; }                             
    }

    public Festival() : base()                                  
    {
        BandsCount = 0;
    }

    public Festival(int id, string name, DateTime date, int bands) : base(id, name, date) 
    {
        BandsCount = bands;
    }

    public override void RegisterParticipant()                  // Полиморфизм: override
    {
        Console.WriteLine($"Регистрация участника на фестиваль с выбором музыкальных жанров. Групп: {bandsCount}");
    }
}

public class Seminar : Event                                    // Наследование от Event
{
    private string topics;                                     

    public string Topics                                        // Своё свойство
    {
        get { return topics; }                                  
        set { topics = value; }                                 
    }

    public Seminar() : base()                                   
    {
        Topics = "Unknown";
    }

    public Seminar(int id, string name, DateTime date, string topics) : base(id, name, date) // Конструктор с параметрами
    {
        Topics = topics;
    }

    public override void GetEventDetails()                      // Полиморфизм: override
    {
        Console.WriteLine($"Семинар: {EventName}, ID: {EventId}, Дата: {Date}, Тематические области: {topics}");
    }
}

public class Program
{
    public static void Main()
    {
        // Создание объектов (экземпляров классов)
        Conference conf = new Conference(1, "IT-конференция", new DateTime(2026, 10, 15), 5);
        Festival fest = new Festival(2, "Рок-фестиваль", new DateTime(2026, 8, 20), 8);
        Seminar sem = new Seminar(3, "Семинар по C#", new DateTime(2026, 9, 10), "ООП, LINQ, async/await");

        conf.RegisterParticipant();                             // Полиморфизм в действии
        fest.RegisterParticipant();                             // Своя версия метода
        sem.RegisterParticipant();                              // Базовая версия метода

        conf.NotifyParticipants();                            
        fest.NotifyParticipants();                              
        sem.NotifyParticipants();                               

        conf.GetEventDetails();                                 
        fest.GetEventDetails();                              
        sem.GetEventDetails();                                  

        conf.InteractWith(sem);                                 // Взаимодействие объектов
        fest.InteractWith(conf);                                
    }
}
```
