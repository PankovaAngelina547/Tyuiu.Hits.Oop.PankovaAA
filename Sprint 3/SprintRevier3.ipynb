```polyglot-notebook
using System;
using System.Collections.Generic;

public interface IInstallable
{
    void Install();
}

public abstract class Component
{
    public string Name { get; set; }
    public decimal Price { get; set; }

    public Component(string name, decimal price)
    {
        Name = name;
        Price = price;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine("Компонент: " + Name + ", Цена: " + Price + " руб.");
    }

    public void GetInfo(string prefix)
    {
        Console.WriteLine("[" + prefix + "] " + Name + " - " + Price + " руб.");
    }

    public override string ToString()
    {
        return Name + " (" + Price + " руб.)";
    }
}

public class CPU : Component, IInstallable
{
    public int Cores { get; set; }

    public CPU(string name, decimal price, int cores) : base(name, price)
    {
        Cores = cores;
    }

    public override void GetInfo()
    {
        Console.WriteLine("Процессор: " + Name + ", Ядра: " + Cores + ", Цена: " + Price + " руб.");
    }

    public void Install()
    {
        Console.WriteLine("Процессор " + Name + " установлен.");
    }

    public static bool operator ==(CPU a, CPU b)
    {
        return a.Name == b.Name && a.Cores == b.Cores;
    }

    public static bool operator !=(CPU a, CPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        if (obj is CPU)
        {
            return this == (CPU)obj;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Cores);
    }
}

public class RAM : Component, IInstallable
{
    public int SizeGb { get; set; }

    public RAM(string name, decimal price, int sizeGb) : base(name, price)
    {
        SizeGb = sizeGb;
    }

    public override void GetInfo()
    {
        Console.WriteLine("ОЗУ: " + Name + ", Объем: " + SizeGb + " ГБ, Цена: " + Price + " руб.");
    }

    public void Install()
    {
        Console.WriteLine("ОЗУ " + Name + " установлено.");
    }

    public static bool operator ==(RAM a, RAM b)
    {
        return a.Name == b.Name && a.SizeGb == b.SizeGb;
    }

    public static bool operator !=(RAM a, RAM b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        if (obj is RAM)
        {
            return this == (RAM)obj;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, SizeGb);
    }
}

public class GPU : Component, IInstallable
{
    public int MemoryGb { get; set; }

    public GPU(string name, decimal price, int memoryGb) : base(name, price)
    {
        MemoryGb = memoryGb;
    }

    public override void GetInfo()
    {
        Console.WriteLine("Видеокарта: " + Name + ", Память: " + MemoryGb + " ГБ, Цена: " + Price + " руб.");
    }

    public void Install()
    {
        Console.WriteLine("Видеокарта " + Name + " установлена.");
    }

    public static bool operator ==(GPU a, GPU b)
    {
        return a.Name == b.Name && a.MemoryGb == b.MemoryGb;
    }

    public static bool operator !=(GPU a, GPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        if (obj is GPU)
        {
            return this == (GPU)obj;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, MemoryGb);
    }
}

public class HDD : Component, IInstallable
{
    public int CapacityGb { get; set; }

    public HDD(string name, decimal price, int capacityGb) : base(name, price)
    {
        CapacityGb = capacityGb;
    }

    public override void GetInfo()
    {
        Console.WriteLine("Жесткий диск: " + Name + ", Объем: " + CapacityGb + " ГБ, Цена: " + Price + " руб.");
    }

    public void Install()
    {
        Console.WriteLine("Диск " + Name + " установлен.");
    }

    public static bool operator ==(HDD a, HDD b)
    {
        return a.Name == b.Name && a.CapacityGb == b.CapacityGb;
    }

    public static bool operator !=(HDD a, HDD b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        if (obj is HDD)
        {
            return this == (HDD)obj;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, CapacityGb);
    }
}

public class Computer
{
    private List<Component> components = new List<Component>();

    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        components.Add(component);
        component.Install();
    }

    public void RemoveComponent<T>(T component) where T : Component
    {
        if (components.Remove(component))
        {
            Console.WriteLine("Компонент " + component.Name + " удален из ПК.");
        }
        else
        {
            Console.WriteLine("Компонент " + component.Name + " не найден.");
        }
    }

    public decimal GetTotalPrice()
    {
        decimal total = 0;
        foreach (var comp in components)
        {
            total += comp.Price;
        }
        return total;
    }

    public void ShowAll()
    {
        Console.WriteLine("\n--- Сборка ПК ---");
        foreach (var comp in components)
        {
            comp.GetInfo();
        }
        Console.WriteLine("Итого: " + GetTotalPrice() + " руб.\n");
    }
}

public static class ComputerFactory
{
public static Computer CreateGamingPC()
{
Computer pc = new Computer();
pc.AddComponent(new CPU("AMD Ryzen 9", 45000m, 12));
pc.AddComponent(new GPU("RTX 4080", 120000m, 16));
pc.AddComponent(new RAM("Corsair 64GB", 18000m, 64));
return pc;
}
}

Console.WriteLine("=== ФАБРИКА ===");
Computer pc = ComputerFactory.CreateGamingPC();
pc.ShowAll();

Console.WriteLine("=== СРАВНЕНИЕ ===");
CPU c1 = new CPU("Intel i5", 22000m, 6);
CPU c2 = new CPU("Intel i5", 22000m, 6);
CPU c3 = new CPU("AMD Ryzen", 20000m, 6);
Console.WriteLine("c1 == c2: " + (c1 == c2));
Console.WriteLine("c1 == c3: " + (c1 == c3));

Console.WriteLine("\n=== МЕТОДЫ ===");
c1.GetInfo();
c1.GetInfo("ЭКСТРЕННО");
Console.WriteLine(c1.ToString());

Console.WriteLine("\n=== УДАЛЕНИЕ ===");
pc.RemoveComponent(new RAM("Corsair 64GB", 18000m, 64));
```
