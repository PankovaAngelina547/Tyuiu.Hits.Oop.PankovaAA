```
using System;
using System.Collections.Generic;

public interface IReportable
{
    void GenerateReport();
}

public class Event
{
    private int eventId;
    private string eventName;
    private DateTime date;
    private int maxParticipants;
    private string location;
    private decimal ticketPrice;
    private static int totalEvents = 0;

    public int EventId { get { return eventId; } set { eventId = value; } }
    public string EventName { get { return eventName; } set { eventName = value; } }
    public DateTime Date { get { return date; } set { date = value; } }
    public int MaxParticipants { get { return maxParticipants; } set { maxParticipants = value; } }
    public string Location { get { return location; } set { location = value; } }
    public decimal TicketPrice { get { return ticketPrice; } set { ticketPrice = value; } }
    public static int TotalEvents { get { return totalEvents; } }

    public Event()
    {
        EventId = 0;
        EventName = "Unknown";
        Date = DateTime.Now;
        MaxParticipants = 0;
        Location = "Unknown";
        TicketPrice = 0;
        totalEvents++;
    }

    public Event(int id, string name, DateTime date)
    {
        EventId = id;
        EventName = name;
        Date = date;
        MaxParticipants = 0;
        Location = "Unknown";
        TicketPrice = 0;
        totalEvents++;
    }

    public virtual void RegisterParticipant()
    {
        Console.WriteLine("Регистрация зрителя на мероприятие");
    }

    public void RegisterParticipant(string participantName)
    {
        Console.WriteLine($"Регистрация зрителя {participantName} на мероприятие {EventName}");
    }

    public void RegisterParticipant(string participantName, int seat)
    {
        Console.WriteLine($"Регистрация {participantName} на место №{seat}");
    }

    public virtual void NotifyParticipants()
    {
        Console.WriteLine("Уведомление зрителей о мероприятии");
    }

    public void NotifyParticipants(string topic)
    {
        Console.WriteLine($"Уведомление по теме '{topic}'");
    }

    public virtual void GetEventDetails()
    {
        Console.WriteLine($"Мероприятие: {eventName}, ID: {eventId}, Дата: {date:dd.MM.yyyy}");
    }

    public void GetEventDetails(bool shortVersion)
    {
        if (shortVersion)
            Console.WriteLine($"Мероприятие: {eventName}");
        else
            GetEventDetails();
    }

    public void InteractWith(Event other)
    {
        Console.WriteLine($"[{EventName}] взаимодействует с [{other.EventName}]");
        Console.WriteLine("   Мероприятия обмениваются участниками.");
    }

    public void ShowCapacity()
    {
        Console.WriteLine($"Максимум зрителей: {maxParticipants}, место: {location}");
    }

    public void ShowTicketPrice()
    {
        Console.WriteLine($"Стоимость билета: {ticketPrice} руб.");
    }
}

public class Concert : Event, IReportable
{
    private int musiciansCount;
    private string genre;
    private bool hasSoundcheck;
    private int setlistCount;

    public int MusiciansCount { get { return musiciansCount; } set { musiciansCount = value; } }
    public string Genre { get { return genre; } set { genre = value; } }
    public bool HasSoundcheck { get { return hasSoundcheck; } set { hasSoundcheck = value; } }
    public int SetlistCount { get { return setlistCount; } set { setlistCount = value; } }

    public Concert() : base()
    {
        MusiciansCount = 0;
        Genre = "Unknown";
        HasSoundcheck = false;
        SetlistCount = 0;
    }

    public Concert(int id, string name, DateTime date, int musicians) : base(id, name, date)
    {
        MusiciansCount = musicians;
        Genre = "Рок";
        HasSoundcheck = true;
        SetlistCount = 15;
    }

    public override void NotifyParticipants()
    {
        Console.WriteLine($"Уведомление зрителей концерта. Музыкантов на сцене: {musiciansCount}, жанр: {genre}");
    }

    public void GenerateReport()
    {
        Console.WriteLine($"Отчёт по концерту: жанр — {genre}, песен в сет-листе — {setlistCount}");
    }

    public void ShowSoundcheck()
    {
        Console.WriteLine($"Саундчек: {(hasSoundcheck ? "будет" : "не будет")}");
    }
}

public class MusicFestival : Event, IReportable
{
    private int bandsCount;
    private string mainGenre;
    private int stagesCount;
    private bool isOutdoor;

    public int BandsCount { get { return bandsCount; } set { bandsCount = value; } }
    public string MainGenre { get { return mainGenre; } set { mainGenre = value; } }
    public int StagesCount { get { return stagesCount; } set { stagesCount = value; } }
    public bool IsOutdoor { get { return isOutdoor; } set { isOutdoor = value; } }

    public MusicFestival() : base()
    {
        BandsCount = 0;
        MainGenre = "Unknown";
        StagesCount = 0;
        IsOutdoor = false;
    }

    public MusicFestival(int id, string name, DateTime date, int bands) : base(id, name, date)
    {
        BandsCount = bands;
        MainGenre = "Рок и Металл";
        StagesCount = 3;
        IsOutdoor = true;
    }

    public override void RegisterParticipant()
    {
        Console.WriteLine($"Регистрация на фестиваль с выбором музыкальных жанров. Всего групп: {bandsCount}");
    }

    public void GenerateReport()
    {
        Console.WriteLine($"Отчёт по фестивалю: главный жанр — {mainGenre}, сцен — {stagesCount}");
    }

    public void ShowStages()
    {
        Console.WriteLine($"Количество сцен: {stagesCount}");
    }

    public void ShowOutdoor()
    {
        Console.WriteLine($"На открытом воздухе: {(isOutdoor ? "да" : "нет")}");
    }
}

public class MasterClass : Event, IReportable
{
    private string topics;
    private string lecturer;
    private int durationHours;
    private bool hasCertificate;

    public string Topics { get { return topics; } set { topics = value; } }
    public string Lecturer { get { return lecturer; } set { lecturer = value; } }
    public int DurationHours { get { return durationHours; } set { durationHours = value; } }
    public bool HasCertificate { get { return hasCertificate; } set { hasCertificate = value; } }

    public MasterClass() : base()
    {
        Topics = "Unknown";
        Lecturer = "Unknown";
        DurationHours = 0;
        HasCertificate = false;
    }

    public MasterClass(int id, string name, DateTime date, string topics) : base(id, name, date)
    {
        Topics = topics;
        Lecturer = "Известный музыкант";
        DurationHours = 3;
        HasCertificate = true;
    }

    public override void GetEventDetails()
    {
        Console.WriteLine($"Мастер-класс: {EventName}, ID: {EventId}, Дата: {Date:dd.MM.yyyy}, Тема: {topics}");
    }

    public void GenerateReport()
    {
        Console.WriteLine($"Отчёт по мастер-классу: ведущий — {lecturer}, длительность — {durationHours} ч.");
    }

    public void ShowLecturer()
    {
        Console.WriteLine($"Ведущий: {lecturer}");
    }

    public void ShowCertificate()
    {
        Console.WriteLine($"Сертификат: {(hasCertificate ? "выдаётся" : "не выдаётся")}");
    }
}

public class OnlineStream : Concert
{
    private string platform;
    private int viewersCount;
    private bool isRecorded;

    public string Platform { get { return platform; } set { platform = value; } }
    public int ViewersCount { get { return viewersCount; } set { viewersCount = value; } }
    public bool IsRecorded { get { return isRecorded; } set { isRecorded = value; } }

    public OnlineStream() : base()
    {
        Platform = "Unknown";
        ViewersCount = 0;
        IsRecorded = false;
    }

    public OnlineStream(int id, string name, DateTime date, int musicians, string platform) : base(id, name, date, musicians)
    {
        Platform = platform;
        ViewersCount = 5000;
        IsRecorded = true;
    }

    public void ShowStreamInfo()
    {
        Console.WriteLine($"Стрим на {Platform}. Зрителей: {ViewersCount}. Запись: {(IsRecorded ? "ведётся" : "нет")}");
    }
}

public class EventCollection<T> where T : Event
{
    private List<T> _events = new List<T>();

    public void Add(T item) { _events.Add(item); }
    public void Remove(T item) { _events.Remove(item); }

    public void DisplayAll()
    {
        foreach (var item in _events)
        {
            item.GetEventDetails();
        }
    }
}

public class Program
{
    public static void Main()
    {
        Console.WriteLine("--- СОЗДАНИЕ МУЗЫКАЛЬНЫХ СОБЫТИЙ ---");
        Concert rockConcert = new Concert(1, "Концерт Linkin Park", new DateTime(2026, 10, 15), 6);
        MusicFestival summerFest = new MusicFestival(2, "Нашествие", new DateTime(2026, 7, 20), 25);
        MasterClass vocalClass = new MasterClass(3, "Уроки вокала", new DateTime(2026, 9, 10), "Дыхание и артикуляция");
        OnlineStream stream = new OnlineStream(4, "Онлайн-акустика", new DateTime(2026, 11, 5), 2, "YouTube");

        rockConcert.Location = "Стадион";
        rockConcert.MaxParticipants = 50000;
        rockConcert.TicketPrice = 3500;

        Console.WriteLine("\n--- РЕГИСТРАЦИЯ (ПЕРЕГРУЗКА) ---");
        rockConcert.RegisterParticipant();
        rockConcert.RegisterParticipant("Алексей");                   // Перегрузка
        rockConcert.RegisterParticipant("Мария", 12);                 // Перегрузка
        summerFest.RegisterParticipant();                             // Переопределение
        vocalClass.RegisterParticipant();
        stream.RegisterParticipant();

        Console.WriteLine("\n--- УВЕДОМЛЕНИЯ (ПОЛИМОРФИЗМ) ---");
        rockConcert.NotifyParticipants();                             // Переопределение
        rockConcert.NotifyParticipants("Новый альбом");               // Перегрузка
        summerFest.NotifyParticipants();
        vocalClass.NotifyParticipants();
        stream.NotifyParticipants();

        Console.WriteLine("\n--- ДЕТАЛИ (ПОЛИМОРФИЗМ) ---");
        rockConcert.GetEventDetails();
        rockConcert.GetEventDetails(true);                            // Перегрузка
        summerFest.GetEventDetails();
        vocalClass.GetEventDetails();                                 // Переопределение
        stream.GetEventDetails();

        Console.WriteLine("\n--- ВЗАИМОДЕЙСТВИЕ ОБЪЕКТОВ ---");
        rockConcert.InteractWith(vocalClass);
        summerFest.InteractWith(rockConcert);
        stream.InteractWith(summerFest);

        Console.WriteLine("\n--- ИНТЕРФЕЙС IReportable ---");
        rockConcert.GenerateReport();
        summerFest.GenerateReport();
        vocalClass.GenerateReport();

        Console.WriteLine("\n--- СПЕЦИФИЧНЫЕ МЕТОДЫ ---");
        rockConcert.ShowCapacity();
        rockConcert.ShowTicketPrice();
        rockConcert.ShowSoundcheck();
        summerFest.ShowStages();
        summerFest.ShowOutdoor();
        vocalClass.ShowLecturer();
        vocalClass.ShowCertificate();
        stream.ShowStreamInfo();

        Console.WriteLine("\n--- GENERIC-КОЛЛЕКЦИЯ ---");
        EventCollection<Event> collection = new EventCollection<Event>();
        collection.Add(rockConcert);
        collection.Add(summerFest);
        collection.Add(vocalClass);
        collection.Add(stream);
        collection.DisplayAll();

        Console.WriteLine($"\n--- СТАТИСТИКА ---");
        Console.WriteLine($"Всего создано мероприятий: {Event.TotalEvents}");
    }
}

Program.Main();
```
