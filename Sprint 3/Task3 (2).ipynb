```
using System;
using System.Collections.Generic;

public interface IReportable
{
    void GenerateReport();
}

public class Event
{
    private int eventId;
    private string eventName;
    private DateTime date;
    private int maxParticipants;
    private string location;
    private decimal ticketPrice;
    private static int totalEvents = 0;

    public int EventId
    {
        get { return eventId; }
        set { eventId = value; }
    }

    public string EventName
    {
        get { return eventName; }
        set { eventName = value; }
    }

    public DateTime Date
    {
        get { return date; }
        set { date = value; }
    }

    public int MaxParticipants
    {
        get { return maxParticipants; }
        set { maxParticipants = value; }
    }

    public string Location
    {
        get { return location; }
        set { location = value; }
    }

    public decimal TicketPrice
    {
        get { return ticketPrice; }
        set { ticketPrice = value; }
    }

    public static int TotalEvents
    {
        get { return totalEvents; }
    }

    public Event()
    {
        EventId = 0;
        EventName = "Unknown";
        Date = DateTime.Now;
        MaxParticipants = 0;
        Location = "Unknown";
        TicketPrice = 0;
        totalEvents++;
    }

    public Event(int id, string name, DateTime date)
    {
        EventId = id;
        EventName = name;
        Date = date;
        MaxParticipants = 0;
        Location = "Unknown";
        TicketPrice = 0;
        totalEvents++;
    }

    public virtual void RegisterParticipant()
    {
        Console.WriteLine("Регистрация участника на мероприятие");
    }

    public void RegisterParticipant(string participantName)
    {
        Console.WriteLine($"Регистрация участника {participantName} на мероприятие {EventName}");
    }

    public void RegisterParticipant(string participantName, int seat)
    {
        Console.WriteLine($"Регистрация {participantName} на место №{seat}");
    }

    public virtual void NotifyParticipants()
    {
        Console.WriteLine("Уведомление участников о мероприятии");
    }

    public void NotifyParticipants(string topic)
    {
        Console.WriteLine($"Уведомление по теме '{topic}'");
    }

    public virtual void GetEventDetails()
    {
        Console.WriteLine($"Мероприятие: {eventName}, ID: {eventId}, Дата: {date}");
    }

    public void GetEventDetails(bool shortVersion)
    {
        if (shortVersion)
            Console.WriteLine($"Мероприятие: {eventName}");
        else
            GetEventDetails();
    }

    public void InteractWith(Event other)
    {
        Console.WriteLine($"[{EventName}] взаимодействует с [{other.EventName}]");
        Console.WriteLine("   Мероприятия обмениваются участниками.");
    }

    public void ShowCapacity()
    {
        Console.WriteLine($"Максимум участников: {maxParticipants}, место: {location}");
    }

    public void ShowTicketPrice()
    {
        Console.WriteLine($"Стоимость билета: {ticketPrice} руб.");
    }
}

public class Conference : Event, IReportable
{
    private int speakersCount;
    private string conferenceTheme;
    private bool hasWorkshops;
    private int sessionCount;

    public int SpeakersCount
    {
        get { return speakersCount; }
        set { speakersCount = value; }
    }

    public string ConferenceTheme
    {
        get { return conferenceTheme; }
        set { conferenceTheme = value; }
    }

    public bool HasWorkshops
    {
        get { return hasWorkshops; }
        set { hasWorkshops = value; }
    }

    public int SessionCount
    {
        get { return sessionCount; }
        set { sessionCount = value; }
    }

    public Conference() : base()
    {
        SpeakersCount = 0;
        ConferenceTheme = "Unknown";
        HasWorkshops = false;
        SessionCount = 0;
    }

    public Conference(int id, string name, DateTime date, int speakers) : base(id, name, date)
    {
        SpeakersCount = speakers;
        ConferenceTheme = "Unknown";
        HasWorkshops = false;
        SessionCount = 0;
    }

    public override void NotifyParticipants()
    {
        Console.WriteLine($"Уведомление участников конференции с программой. Спикеров: {speakersCount}");
    }

    public void GenerateReport()
    {
        Console.WriteLine($"Отчёт по конференции: тема — {conferenceTheme}, сессий — {sessionCount}");
    }

    public void ShowWorkshops()
    {
        Console.WriteLine($"Воркшопы: {(hasWorkshops ? "есть" : "нет")}");
    }

    public void ShowTheme()
    {
        Console.WriteLine($"Тема конференции: {conferenceTheme}");
    }
}

public class Festival : Event, IReportable
{
    private int bandsCount;
    private string genre;
    private int stagesCount;
    private bool isOutdoor;

    public int BandsCount
    {
        get { return bandsCount; }
        set { bandsCount = value; }
    }

    public string Genre
    {
        get { return genre; }
        set { genre = value; }
    }

    public int StagesCount
    {
        get { return stagesCount; }
        set { stagesCount = value; }
    }

    public bool IsOutdoor
    {
        get { return isOutdoor; }
        set { isOutdoor = value; }
    }

    public Festival() : base()
    {
        BandsCount = 0;
        Genre = "Unknown";
        StagesCount = 0;
        IsOutdoor = false;
    }

    public Festival(int id, string name, DateTime date, int bands) : base(id, name, date)
    {
        BandsCount = bands;
        Genre = "Unknown";
        StagesCount = 0;
        IsOutdoor = false;
    }

    public override void RegisterParticipant()
    {
        Console.WriteLine($"Регистрация участника на фестиваль с выбором музыкальных жанров. Групп: {bandsCount}");
    }

    public void GenerateReport()
    {
        Console.WriteLine($"Отчёт по фестивалю: жанр — {genre}, сцен — {stagesCount}");
    }

    public void ShowStages()
    {
        Console.WriteLine($"Количество сцен: {stagesCount}");
    }

    public void ShowOutdoor()
    {
        Console.WriteLine($"На открытом воздухе: {(isOutdoor ? "да" : "нет")}");
    }
}

public class Seminar : Event, IReportable
{
    private string topics;
    private string lecturer;
    private int durationHours;
    private bool hasCertificate;

    public string Topics
    {
        get { return topics; }
        set { topics = value; }
    }

    public string Lecturer
    {
        get { return lecturer; }
        set { lecturer = value; }
    }

    public int DurationHours
    {
        get { return durationHours; }
        set { durationHours = value; }
    }

    public bool HasCertificate
    {
        get { return hasCertificate; }
        set { hasCertificate = value; }
    }

    public Seminar() : base()
    {
        Topics = "Unknown";
        Lecturer = "Unknown";
        DurationHours = 0;
        HasCertificate = false;
    }

    public Seminar(int id, string name, DateTime date, string topics) : base(id, name, date)
    {
        Topics = topics;
        Lecturer = "Unknown";
        DurationHours = 0;
        HasCertificate = false;
    }

    public override void GetEventDetails()
    {
        Console.WriteLine($"Семинар: {EventName}, ID: {EventId}, Дата: {Date}, Тематические области: {topics}");
    }

    public void GenerateReport()
    {
        Console.WriteLine($"Отчёт по семинару: лектор — {lecturer}, длительность — {durationHours} ч.");
    }

    public void ShowLecturer()
    {
        Console.WriteLine($"Лектор: {lecturer}");
    }

    public void ShowCertificate()
    {
        Console.WriteLine($"Сертификат: {(hasCertificate ? "выдаётся" : "не выдаётся")}");
    }
}

public class OnlineConference : Conference
{
    private string platform;
    private int viewersCount;
    private bool isRecorded;

    public string Platform
    {
        get { return platform; }
        set { platform = value; }
    }

    public int ViewersCount
    {
        get { return viewersCount; }
        set { viewersCount = value; }
    }

    public bool IsRecorded
    {
        get { return isRecorded; }
        set { isRecorded = value; }
    }

    public OnlineConference() : base()
    {
        Platform = "Unknown";
        ViewersCount = 0;
        IsRecorded = false;
    }

    public OnlineConference(int id, string name, DateTime date, int speakers, string platform) : base(id, name, date, speakers)
    {
        Platform = platform;
        ViewersCount = 0;
        IsRecorded = false;
    }

    public void ShowRecording()
    {
        Console.WriteLine($"Запись: {(isRecorded ? "ведётся" : "не ведётся")}");
    }
}

public class EventCollection<T> where T : Event
{
    private List<T> _events = new List<T>();

    public void Add(T item)
    {
        _events.Add(item);
    }

    public void Remove(T item)
    {
        _events.Remove(item);
    }

    public void DisplayAll()
    {
        foreach (var item in _events)
        {
            item.GetEventDetails();
        }
    }
}

public class Program
{
    public static void Main()
    {
        Conference conf = new Conference(1, "IT-конференция", new DateTime(2026, 10, 15), 5);
        Festival fest = new Festival(2, "Рок-фестиваль", new DateTime(2026, 8, 20), 8);
        Seminar sem = new Seminar(3, "Семинар по C#", new DateTime(2026, 9, 10), "ООП, LINQ, async/await");
        OnlineConference online = new OnlineConference(4, "Онлайн-митап", new DateTime(2026, 11, 5), 3, "Zoom");

        conf.RegisterParticipant();
        conf.RegisterParticipant("Иван");
        conf.RegisterParticipant("Мария", 12);
        fest.RegisterParticipant();
        sem.RegisterParticipant();
        online.RegisterParticipant();

        conf.NotifyParticipants();
        conf.NotifyParticipants("AI");
        fest.NotifyParticipants();
        sem.NotifyParticipants();
        online.NotifyParticipants();

        conf.GetEventDetails();
        conf.GetEventDetails(true);
        fest.GetEventDetails();
        sem.GetEventDetails();
        online.GetEventDetails();

        conf.InteractWith(sem);
        fest.InteractWith(conf);
        online.InteractWith(fest);

        conf.GenerateReport();
        fest.GenerateReport();
        sem.GenerateReport();

        conf.ShowCapacity();
        conf.ShowTicketPrice();
        conf.ShowTheme();
        conf.ShowWorkshops();
        fest.ShowStages();
        fest.ShowOutdoor();
        sem.ShowLecturer();
        sem.ShowCertificate();
        online.ShowRecording();

        EventCollection<Event> collection = new EventCollection<Event>();
        collection.Add(conf);
        collection.Add(fest);
        collection.Add(sem);
        collection.Add(online);
        collection.DisplayAll();

        Console.WriteLine($"Всего мероприятий: {Event.TotalEvents}");
    }
}
```
