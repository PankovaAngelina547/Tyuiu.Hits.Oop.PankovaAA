```
using System;

public interface IReportable                                
{
    void GenerateReport();
}

public class Event                                          // Базовый класс — простое наследование
{
    private int eventId;
    private string eventName;
    private DateTime date;
    private int maxParticipants;                            // Новый атрибут: лимит участников
    private string location;                                // Новый атрибут: место проведения
    private decimal ticketPrice;                            // Новый атрибут: цена билета
    private static int totalEvents = 0;                     // Статический счётчик всех мероприятий

    public int EventId
    {
        get { return eventId; }
        set { eventId = value; }
    }

    public string EventName
    {
        get { return eventName; }
        set { eventName = value; }
    }

    public DateTime Date
    {
        get { return date; }
        set { date = value; }
    }

    public int MaxParticipants                              // Свойство для лимита участников
    {
        get { return maxParticipants; }
        set { maxParticipants = value; }
    }

    public string Location                                  // Свойство для места
    {
        get { return location; }
        set { location = value; }
    }

    public decimal TicketPrice                              // Свойство для цены
    {
        get { return ticketPrice; }
        set { ticketPrice = value; }
    }

    public static int TotalEvents                           // Статическое свойство — общий счётчик
    {
        get { return totalEvents; }
    }

    public Event()
    {
        EventId = 0;
        EventName = "Unknown";
        Date = DateTime.Now;
        MaxParticipants = 0;
        Location = "Unknown";
        TicketPrice = 0;
        totalEvents++;                                      
    }

    public Event(int id, string name, DateTime date)
    {
        EventId = id;
        EventName = name;
        Date = date;
        MaxParticipants = 0;
        Location = "Unknown";
        TicketPrice = 0;
        totalEvents++;                                    
    }

    public virtual void RegisterParticipant()               // Полиморфизм: переопределяется в Festival
    {
        Console.WriteLine("Регистрация участника на мероприятие");
    }

    public virtual void NotifyParticipants()                // Полиморфизм: переопределяется в Conference
    {
        Console.WriteLine("Уведомление участников о мероприятии");
    }

    public virtual void GetEventDetails()                   // Полиморфизм: переопределяется в Seminar
    {
        Console.WriteLine($"Мероприятие: {eventName}, ID: {eventId}, Дата: {date}");
    }

    public void InteractWith(Event other)                   // Взаимодействие объектов
    {
        Console.WriteLine($"[{EventName}] взаимодействует с [{other.EventName}]");
        Console.WriteLine("   Мероприятия обмениваются участниками.");
    }

    public void ShowCapacity()                              // Новый метод: вывод лимита и места
    {
        Console.WriteLine($"Максимум участников: {maxParticipants}, место: {location}");
    }

    public void ShowTicketPrice()                           // Новый метод: вывод цены
    {
        Console.WriteLine($"Стоимость билета: {ticketPrice} руб.");
    }
}

public class Conference : Event, IReportable                // Простое наследование + интерфейс
{
    private int speakersCount;
    private string conferenceTheme;                         // Новый атрибут: тема конференции
    private bool hasWorkshops;                              // Новый атрибут: наличие воркшопов
    private int sessionCount;                               // Новый атрибут: число сессий

    public int SpeakersCount
    {
        get { return speakersCount; }
        set { speakersCount = value; }
    }

    public string ConferenceTheme                          
    {
        get { return conferenceTheme; }
        set { conferenceTheme = value; }
    }

    public bool HasWorkshops                                
    {
        get { return hasWorkshops; }
        set { hasWorkshops = value; }
    }

    public int SessionCount                                
    {
        get { return sessionCount; }
        set { sessionCount = value; }
    }

    public Conference() : base()
    {
        SpeakersCount = 0;
        ConferenceTheme = "Unknown";
        HasWorkshops = false;
        SessionCount = 0;
    }

    public Conference(int id, string name, DateTime date, int speakers) : base(id, name, date)
    {
        SpeakersCount = speakers;
        ConferenceTheme = "Unknown";
        HasWorkshops = false;
        SessionCount = 0;
    }

    public override void NotifyParticipants()               
    {
        Console.WriteLine($"Уведомление участников конференции с программой. Спикеров: {speakersCount}");
    }

    public void GenerateReport()                           
    {
        Console.WriteLine($"Отчёт по конференции: тема — {conferenceTheme}, сессий — {sessionCount}");
    }

    public void ShowWorkshops()                             
    {
        Console.WriteLine($"Воркшопы: {(hasWorkshops ? "есть" : "нет")}");
    }

    public void ShowTheme()                                
    {
        Console.WriteLine($"Тема конференции: {conferenceTheme}");
    }
}

public class Festival : Event, IReportable
{
    private int bandsCount;
    private string genre;                                   // Новый атрибут: жанр
    private int stagesCount;                                // Новый атрибут: число сцен
    private bool isOutdoor;                                 // Новый атрибут: на улице ли

    public int BandsCount
    {
        get { return bandsCount; }
        set { bandsCount = value; }
    }

    public string Genre                                     // Свойство жанра
    {
        get { return genre; }
        set { genre = value; }
    }

    public int StagesCount                                  // Свойство сцен
    {
        get { return stagesCount; }
        set { stagesCount = value; }
    }

    public bool IsOutdoor                                   // Свойство "на улице"
    {
        get { return isOutdoor; }
        set { isOutdoor = value; }
    }

    public Festival() : base()
    {
        BandsCount = 0;
        Genre = "Unknown";
        StagesCount = 0;
        IsOutdoor = false;
    }

    public Festival(int id, string name, DateTime date, int bands) : base(id, name, date)
    {
        BandsCount = bands;
        Genre = "Unknown";
        StagesCount = 0;
        IsOutdoor = false;
    }

    public override void RegisterParticipant()              // Переопределение базового метода
    {
        Console.WriteLine($"Регистрация участника на фестиваль с выбором музыкальных жанров. Групп: {bandsCount}");
    }

    public void GenerateReport()                            // Реализация контракта IReportable
    {
        Console.WriteLine($"Отчёт по фестивалю: жанр — {genre}, сцен — {stagesCount}");
    }

    public void ShowStages()                                // Новый метод
    {
        Console.WriteLine($"Количество сцен: {stagesCount}");
    }

    public void ShowOutdoor()                               // Новый метод
    {
        Console.WriteLine($"На открытом воздухе: {(isOutdoor ? "да" : "нет")}");
    }
}

public class Seminar : Event, IReportable
{
    private string topics;
    private string lecturer;                                // Новый атрибут: лектор
    private int durationHours;                              // Новый атрибут: длительность
    private bool hasCertificate;                            // Новый атрибут: сертификат

    public string Topics
    {
        get { return topics; }
        set { topics = value; }
    }

    public string Lecturer                                  // Свойство лектора
    {
        get { return lecturer; }
        set { lecturer = value; }
    }

    public int DurationHours                                // Свойство длительности
    {
        get { return durationHours; }
        set { durationHours = value; }
    }

    public bool HasCertificate                              // Свойство сертификата
    {
        get { return hasCertificate; }
        set { hasCertificate = value; }
    }

    public Seminar() : base()
    {
        Topics = "Unknown";
        Lecturer = "Unknown";
        DurationHours = 0;
        HasCertificate = false;
    }

    public Seminar(int id, string name, DateTime date, string topics) : base(id, name, date)
    {
        Topics = topics;
        Lecturer = "Unknown";
        DurationHours = 0;
        HasCertificate = false;
    }

    public override void GetEventDetails()                  // Переопределение базового метода
    {
        Console.WriteLine($"Семинар: {EventName}, ID: {EventId}, Дата: {Date}, Тематические области: {topics}");
    }

    public void GenerateReport()                            // Реализация контракта IReportable
    {
        Console.WriteLine($"Отчёт по семинару: лектор — {lecturer}, длительность — {durationHours} ч.");
    }

    public void ShowLecturer()                              // Новый метод
    {
        Console.WriteLine($"Лектор: {lecturer}");
    }

    public void ShowCertificate()                           // Новый метод
    {
        Console.WriteLine($"Сертификат: {(hasCertificate ? "выдаётся" : "не выдаётся")}");
    }
}

public class OnlineConference : Conference                  // Сложное наследование: от производного класса
{
    private string platform;                                // Новый атрибут: платформа
    private int viewersCount;                               // Новый атрибут: число зрителей
    private bool isRecorded;                                // Новый атрибут: ведётся ли запись

    public string Platform                                  // Свойство платформы
    {
        get { return platform; }
        set { platform = value; }
    }

    public int ViewersCount                                 // Свойство зрителей
    {
        get { return viewersCount; }
        set { viewersCount = value; }
    }

    public bool IsRecorded                                  // Свойство записи
    {
        get { return isRecorded; }
        set { isRecorded = value; }
    }

    public OnlineConference() : base()
    {
        Platform = "Unknown";
        ViewersCount = 0;
        IsRecorded = false;
    }

    public OnlineConference(int id, string name, DateTime date, int speakers, string platform) : base(id, name, date, speakers)
    {
        Platform = platform;
        ViewersCount = 0;
        IsRecorded = false;
    }

    public void ShowRecording()                             // Новый метод
    {
        Console.WriteLine($"Запись: {(isRecorded ? "ведётся" : "не ведётся")}");
    }
}

public class Program
{
    public static void Main()
    {
        Conference conf = new Conference(1, "IT-конференция", new DateTime(2026, 10, 15), 5);
        Festival fest = new Festival(2, "Рок-фестиваль", new DateTime(2026, 8, 20), 8);
        Seminar sem = new Seminar(3, "Семинар по C#", new DateTime(2026, 9, 10), "ООП, LINQ, async/await");
        OnlineConference online = new OnlineConference(4, "Онлайн-митап", new DateTime(2026, 11, 5), 3, "Zoom"); // Сложное наследование

        conf.RegisterParticipant();
        fest.RegisterParticipant();                         // Полиморфизм — своя версия
        sem.RegisterParticipant();
        online.RegisterParticipant();

        conf.NotifyParticipants();                          // Полиморфизм — своя версия
        fest.NotifyParticipants();
        sem.NotifyParticipants();
        online.NotifyParticipants();

        conf.GetEventDetails();
        fest.GetEventDetails();
        sem.GetEventDetails();                              // Полиморфизм — своя версия
        online.GetEventDetails();

        conf.InteractWith(sem);                             // Взаимодействие объектов
        fest.InteractWith(conf);
        online.InteractWith(fest);

        conf.GenerateReport();                              // Множественное наследование через интерфейс IReportable
        fest.GenerateReport();
        sem.GenerateReport();

        conf.ShowCapacity();                                // Новые методы базового класса
        conf.ShowTicketPrice();
        conf.ShowTheme();                                   // Новые методы Conference
        conf.ShowWorkshops();
        fest.ShowStages();                                  // Новые методы Festival
        fest.ShowOutdoor();
        sem.ShowLecturer();                                 // Новые методы Seminar
        sem.ShowCertificate();
        online.ShowRecording();                             // Новый метод OnlineConference

        Console.WriteLine($"Всего мероприятий: {Event.TotalEvents}");  // Статический счётчик
    }
}
```
