```polyglot-notebook
using System;
using System.Collections.Generic;

public class Product
{
    private string name;
    private decimal price;
    private int quantity;
    protected string category;
    internal string description;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public decimal Price
    {
        get { return price; }
        set { price = value; }
    }

    public int Quantity
    {
        get { return quantity; }
        set { quantity = value; }
    }

    public Product()
    {
        Name = "Unknown";
        Price = 0;
        Quantity = 0;
        category = "General";
        description = "No description";
    }

    public Product(string name, decimal price, int quantity)
    {
        Name = name;
        Price = price;
        Quantity = quantity;
        category = "General";
        description = "No description";
    }

    public decimal GetTotalPrice()
    {
        return Price * Quantity;
    }

    public void UpdateQuantity(int amount)
    {
        Quantity += amount;
    }

    public string GetProductInfo()
    {
        return $"Товар: {Name}, Цена: {Price}, Количество: {Quantity}, Категория: {category}, Описание: {description}";
    }
}

public class Seller
{
    private string name;
    private string employeeId;
    protected decimal salary;
    internal string contactInfo;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public string EmployeeId
    {
        get { return employeeId; }
        set { employeeId = value; }
    }

    public Seller()
    {
        Name = "Unknown";
        EmployeeId = "0000";
        salary = 0;
        contactInfo = "No contact";
    }

    public Seller(string name, string employeeId, decimal salary)
    {
        Name = name;
        EmployeeId = employeeId;
        this.salary = salary;
        contactInfo = "No contact";
    }

    public void AddProduct(Product product)
    {
        Console.WriteLine($"Продавец {Name} добавил товар: {product.Name}");
    }

    public void SellProduct(Product product, int quantity)
    {
        product.UpdateQuantity(-quantity);
        Console.WriteLine($"Продавец {Name} продал {quantity} шт. товара {product.Name}");
    }

    public string GetSellerInfo()
    {
        return $"Продавец: {Name}, ID: {EmployeeId}, Зарплата: {salary}, Контакты: {contactInfo}";
    }
}

public class Store
{
    private string storeName;
    private string location;
    public string storeHours;

    private List<Product> products = new List<Product>();

    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }

    public string Location
    {
        get { return location; }
        set { location = value; }
    }

    public Store()
    {
        StoreName = "Unknown";
        Location = "Unknown";
        storeHours = "9:00 - 18:00";
    }

    public Store(string storeName, string location, string storeHours)
    {
        StoreName = storeName;
        Location = location;
        this.storeHours = storeHours;
    }

    public void AddSeller(Seller seller)
    {
        Console.WriteLine($"Магазин {StoreName} принял на работу продавца: {seller.Name}");
    }

    public void ListProducts()
    {
        Console.WriteLine($"Товары в магазине {StoreName}:");
        if (products.Count == 0)
        {
            Console.WriteLine("  (пусто)");
        }
        else
        {
            foreach (var p in products)
            {
                Console.WriteLine($"  - {p.GetProductInfo()}");
            }
        }
    }

    public string GetStoreInfo()
    {
        return $"Магазин: {StoreName}, Адрес: {Location}, Часы работы: {storeHours}";
    }

    public void AddProductToStore(Product product)
    {
        products.Add(product);
    }
}

// --- ЗАПУСК КОДА (без Main) ---
Product product = new Product("Ноутбук", 75000.50m, 10);
product.description = "Игровой ноутбук";

Seller seller = new Seller("Иван Петров", "EMP-001", 50000m);
seller.contactInfo = "ivan@mail.com";

Store store = new Store("ТехноМир", "ул. Ленина, 15", "10:00 - 21:00");

store.AddSeller(seller);
store.AddProductToStore(product);
seller.AddProduct(product);
seller.SellProduct(product, 2);

Console.WriteLine(store.GetStoreInfo());
Console.WriteLine(seller.GetSellerInfo());
Console.WriteLine(product.GetProductInfo());
Console.WriteLine($"Общая стоимость: {product.GetTotalPrice()}");
store.ListProducts();
```
